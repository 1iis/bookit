# Bookit
> A runtime for hyperbooks.

## Core model

`bookit` is the runtime, CLI, library, and host integration layer for **hyperbooks**.

A **hyperbook** is a portable software-backed book: content, assets, metadata, optional code, optional interfaces, and optional reader state rules. It may be a course, interactive textbook, rich media album, fiction work, documentary, game-like experience, technical manual, or other programmable media object.

### Core terms

| Term | Meaning |
|---|---|
| `bookit` | Runtime/tooling that loads, runs, inspects, and manages hyperbooks. |
| hyperbook | The medium/object being run. |
| `hyperbook.toml` | Manifest describing one hyperbook. |
| host | Environment where `bookit` runs: Solveit, local PC, VPS, container, server, etc. |
| surface | UX through which a hyperbook is experienced: CLI, web, Solveit, notebook, dashboard, game view, etc. |

The console/cartridge metaphor is useful but informal: `bookit` is like the console; a hyperbook is like the cartridge. Technically, a hyperbook may be a folder, Git repo, archive, or later a one-file bundle.

## North star

`bookit` lets people load, run, inspect, modify, fork, share, and remove hyperbooks.

The first UX target is:

```bash
bookit load https://github.com/1iis/bookofcomputers
bookit run bookofcomputers
```

or, inside a local hyperbook folder:

```bash
bookit run .
```

## MVP focus

The first proof should be intentionally tiny:

```bash
bookit inspect examples/minimal-hyperbook
bookit run examples/minimal-hyperbook
```

Minimum behavior:

1. recognize a folder containing `hyperbook.toml`;
2. parse and validate the manifest;
3. inspect the hyperbook in a readable way;
4. run the default Markdown entry;
5. keep the implementation small enough to ship quickly.

No registry, bundle format, dependency solver, installer magic, web UI, marketplace, or rich runtime until the local inspect/run loop works.

## Design constraints

- Declarative first: `hyperbook.toml` is the first contract.
- Code when earned: executable hooks must be visible, intentional, and capability-aware.
- Source and reader state stay separate.
- Hyperbooks should be portable and inspectable.
- The runtime owns host integration; each hyperbook should not reinvent installation logic.
- Prefer stdlib and small obvious interfaces.
- Avoid blind `curl | sh` culture as the normal trust model.
- `bookit` must remain hyperbook-agnostic. The Book of Computers is first dogfood, not a special case.

## Implementation map

### Initial module intent

| Module | Responsibility |
|---|---|
| `manifest.py` | Load, parse, validate, and represent `hyperbook.toml`. |
| `paths.py` | Resolve repo, hyperbook, library, source, and state paths. |
| `library.py` | Register/load/list/eject local hyperbooks. |
| `cli.py` | Thin command-line interface over the library/runtime functions. |

### Initial tests

| Test file | Responsibility |
|---|---|
| `tests/test_manifest.py` | Manifest parsing and validation. |
| `tests/test_library.py` | Local hyperbook loading/running behavior. |

### Initial example

| Example | Purpose |
|---|---|
| `examples/minimal-hyperbook/` | Smallest valid hyperbook: manifest + README + Markdown entry. |

## Workflow

Work from `META.ipynb` first. Use `DEV` dialogs only when a feature needs narrower implementation context.

This repo should follow the 1iis Solveit + nbdev workflow:

1. plan in `META` or `DEV`;
2. write durable literate source notebooks under `nbs/`;
3. export package modules with `nbdev-export`;
4. treat `src/bookit/*.py` as generated artifacts;
5. test the real package with editable install and autoreload.

Keep CRAFT updated only with stable decisions needed on every prompt. Put detailed reasoning, transient plans, and implementation sequencing in `META.ipynb`.

Design pressure:

> Avoid the framework cathedral. Ship the tiny loop first.
