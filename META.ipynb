## Handoff from BoC META
> Continuity note for starting `bookit` from the Book of Computers control-plane work.

This section transfers only the `bookit`-relevant context from [📑`boc/META`](dialog_?name=1iis%2Fboc%2FMETA). It is not a transcript. It is the minimum project memory needed for a synth to continue work here as if the design conversation had happened in this repo.

`bookit` emerged from the Book of Computers (BoC) planning work as the generic runtime/tooling layer for **hyperbooks**. The Book of Computers is expected to be the first serious dogfood hyperbook, but `bookit` must stay hyperbook-agnostic.

### Core decision

Use this naming model:

| Name | Meaning |
|---|---|
| `bookit` | Runtime, CLI, library, and host integration layer. |
| hyperbook | Portable software-backed book/media object being run. |
| `hyperbook.toml` | Manifest describing one hyperbook. |
| host | Environment where `bookit` runs: Solveit, local PC, VPS, container, server, etc. |
| surface | UX for experiencing a hyperbook: CLI, web, Solveit, notebook, dashboard, game view, etc. |

Do **not** say `bookit` runs “bookits.”  
`bookit` runs **hyperbooks**.

The console/cartridge metaphor is useful but informal:

> `bookit` is like the console.  
> A hyperbook is like the cartridge.

Do not over-formalize “cartridge” in technical APIs yet. It is a memory aid for portability, ownability, loading/ejecting, and physical-media-like UX.

### Product shape

A hyperbook is a coherent portable software object: content, assets, metadata, optional code, optional interfaces, and optional reader state rules.

It may be:

- a course;
- an interactive textbook;
- a rich media album;
- fiction with programmable structure;
- a documentary;
- a game-like experience;
- a technical manual;
- a living research notebook;
- any other inspectable programmable media object.

The long arc is **content as software**: a hyperbook can support operations that ordinary static media cannot, such as renaming a character across fiction, reordering an album, adding synth explanations to every paragraph, generating alternate views, attaching tests to lessons, or turning content into dashboards and tools.

Do not try to implement the whole long arc now. Preserve the possibility.

### North-star UX

Endgame shape:

```bash
bookit load https://github.com/1iis/bookofcomputers
bookit run bookofcomputers
```

Local development shape:

```bash
bookit run .
```

First proof shape:

```bash
bookit inspect examples/minimal-hyperbook
bookit run examples/minimal-hyperbook
```

The exact command vocabulary may change. The shape should not: get runtime, load/recognize hyperbook, inspect it, run it, preserve state separately, allow deeper inspection/forking later.

### MVP boundary

First milestone:

> A local folder with `hyperbook.toml` is recognized as a hyperbook, can be inspected, and can be run through `bookit`.

Minimum behavior:

1. define a minimal `hyperbook.toml`;
2. parse it with stdlib `tomllib`;
3. validate required fields;
4. resolve entry path;
5. inspect a hyperbook in readable CLI output;
6. run the default Markdown entry in the simplest useful way;
7. test the behavior against `examples/minimal-hyperbook`.

Do not build yet:

- registry;
- marketplace;
- bundle format;
- dependency solver;
- arbitrary installer system;
- web UI;
- Solveit integration;
- rich state engine;
- permission/capability framework beyond what is needed to keep the design honest.

### Design constraints

Keep these invariant unless deliberately revised:

1. Declarative first: `hyperbook.toml` is the first contract.
2. Code when earned: executable hooks must be visible, intentional, and capability-aware.
3. Source and reader state stay separate.
4. Hyperbooks should be portable and inspectable.
5. Runtime owns host integration; hyperbooks describe needs, hosts decide how to satisfy them.
6. Prefer stdlib and small obvious interfaces.
7. Avoid blind `curl | sh` as the normal trust model.
8. Bundle/archive formats are later convenience formats, not the essence.
9. BoC is dogfood, not a special case.
10. Ship the tiny loop before growing the framework.

### Initial implementation map

Likely package module responsibilities:

| Module | Responsibility |
|---|---|
| `manifest.py` | Load, parse, validate, and represent `hyperbook.toml`. |
| `paths.py` | Resolve repo, hyperbook, library, source, entry, and state paths. |
| `library.py` | Register/load/list/eject local hyperbooks; initially may stay tiny. |
| `cli.py` | Thin command-line interface over manifest/library/runtime functions. |

Likely first tests:

| Test | Purpose |
|---|---|
| `tests/test_manifest.py` | Valid manifest parsing, missing required fields, entry resolution. |
| `tests/test_library.py` | Local example hyperbook can be inspected/run through package functions. |
